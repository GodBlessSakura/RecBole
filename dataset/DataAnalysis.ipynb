```python
# 读取原有文件进行数据分析
import pandas as pd
origin_file_path = '/tmp/pycharm_project_946/dataset/CiteULike/CiteULike/all.csv'
df = pd.read_csv(origin_file_path)
```

```python
df.shape
```

```python
df
```

```python
user_counts = df['user'].value_counts()
item_counts = df['item'].value_counts()

# 过滤掉交互次数不足10次的user和item
filtered_df = df[df['user'].isin(user_counts[user_counts >= 10].index) & df['item'].isin(item_counts[item_counts >= 10].index)]

# 返回过滤后的df
print(filtered_df)
```

```python
filtered_df = filtered_df.reset_index(drop=True)
```

```python
filtered_df
```

```python
# 对 user 列进行重新编码
filtered_df['user'] = pd.factorize(filtered_df['user'])[0]

# 对 item 列进行重新编码
filtered_df['item'] = pd.factorize(filtered_df['item'])[0]

# 返回重新编码后的df
print(filtered_df)
```

```python
filtered_df.to_csv('/tmp/pycharm_project_946/dataset/CiteULike/CiteULike/CiteULike.inter', index=False)
```
